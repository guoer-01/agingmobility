```python
#read data
listParentType=['Health Care Services',
                 'Pharmaceutical and Medical Supply Stores',
                 'Grocery and Food Supply',
                 'Social Assistance',
                 'Transportation Services',
                 'Home Health Care Services',
                 'Financial Services',
                 'Legal Services',
                 'Housing and Real Estate',
                 'Personal Care Services',
                 'Utility Services',
                 'Retail Stores for Basic Needs']
allList=[]

###data source 3: accratios
for key in listParentType: 
    homePath='data/A under 5-mile setting/'+key+'.csv'
    df1=pd.read_csv(homePath,usecols=['census_block_group','ParentType','demand_visits','demand_ratios'])
    allList.append(df1)
    print(key)
###data source 3: accratios

dfAll=pd.concat(allList)   
```
